```python
import numpy as np
import pandas as pd
```

```python
L = pd.DataFrame([
    [0, 0, 1, 0],
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 0, 0, 0]
])
d = 0.85

E = pd.DataFrame(np.ones([4, 4]) / 4)
D = np.diag(np.sum(L, axis=0))
D_inv = np.linalg.inv(D)
A = d * L @ D_inv  + (1 - d) * E
p0 = np.ones(4)
```

```python
A
```

```python
p = p0.copy()
for i in range(100):
    p = A @ p
    p = p / p.sum()
p
```
